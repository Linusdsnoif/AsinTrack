```python
import pandas as pd
import re

file_path = "raw.xlsx"

# Read the specific sheet, skipping the top 4 metadata rows
df = pd.read_excel(file_path, sheet_name="Tray auto lesley", header=4, dtype=str)

# The first column is "Row Labels"
row_labels = df["Row Labels"].dropna()

# Amazon ASIN pattern: starts with B + 9 alphanumeric chars (case-insensitive)
ASIN_PATTERN = re.compile(r'^[Bb][0-9A-Za-z]{9}$')

asins = []
for value in row_labels:
    value = value.strip()
    if ASIN_PATTERN.match(value):
        asins.append(value.upper())


print(f"Total ASINs to save: {len(asins)}")
for asin in asins:
    print(asin)

# Save to Excel without the header row
output_df = pd.DataFrame(asins)
output_df.to_excel("tray_asins.xlsx", index=False, header=False)
print("\nSaved to tray_asins.xlsx")
```
